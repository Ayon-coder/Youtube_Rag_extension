```python
import requests
import yt_dlp

url = "https://www.youtube.com/watch?v=PRp5Y543LN0"

with yt_dlp.YoutubeDL({"skip_download": True}) as ydl:
    info = ydl.extract_info(url, download=False)

captions = info.get("subtitles") or info.get("automatic_captions")

if not captions or "en" not in captions:
    print("No English captions found")
    exit()

caption_url = captions["en"][0]["url"]

text = requests.get(caption_url).text

print(text[:500])
```

```python
caption_url = next(
    x["url"] for x in captions["en"] if x["ext"] == "json3"
)
```

```python
import requests

data = requests.get(caption_url).json()

transcript = " ".join(
    seg["utf8"]
    for event in data["events"]
    for seg in event.get("segs", [])
)

print(transcript)
```
